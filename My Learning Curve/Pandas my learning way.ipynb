# Data Cleaning in Pandas

```python
import pandas as pd
```

```python
df = pd.read_excel(r"D:\Data_Analytics\Python\PandasYouTubeSeries-main\Customer Call List.xlsx")
df
```

```python
df = df.drop_duplicates()
df
```

```python
# df.drop(columns = 'Not_Useful_Column', inplace = True) 
df = df.drop(columns = 'Not_Useful_Column')
df
```

```python
#df["Last_Name"] = df["Last_Name"].str.lstrip("...")
#df["Last_Name"] = df["Last_Name"].str.lstrip("/")
#df["Last_Name"] = df["Last_Name"].str.rstrip("_")
df["Last_Name"] = df["Last_Name"].str.strip("123._/")
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace(r'[^a-zA-Z0-9]','', regex=True ) # ми беремо все, що не від а до z, від A до Z, від 0 до 9 та замінюємо на "нічого"
df["Phone_Number"]
```

```python
# for x in df["Phone_Number"]:
#     x = str(x)
df["Phone_Number"] = df["Phone_Number"].apply(lambda x: str(x))
df["Phone_Number"] = df["Phone_Number"].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10])
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace('nan--', '', regex=True)
df["Phone_Number"] = df["Phone_Number"].str.replace('Na--', '', regex=True)
df
```

```python
df[["Street_Adress", "State", "Zip_Code"]] = df['Address'].str.split(',', expand = True)
df
```

```python
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('Yes', 'Y')
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('No', 'N')
df
```

```python
#df = df.replace('N/a', '')
#df = df.replace('NaN', '')
```

```python
df=df.fillna('')
df
```

```python
for x in df.index:
    if df.loc[x, "Do_Not_Contact"] == 'Y':
        df.drop(x, inplace = True)
df
```

```python
for x in df.index:
    if df.loc[x, "Phone_Number"] == '':
        df.drop(x, inplace = True)
df
####
# df.dropna(subset="Phone_Number", inplace = True)  another way to drop values
```

```python
df = df.reset_index(drop = True)
df
```
