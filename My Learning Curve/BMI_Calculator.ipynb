# BMI Calculation

```python
#BMI = (weight in kg)/(height square in meters)
```

```python
Classification	BMI range - kg/m2
Severe Thinness	< 16
Moderate Thinness	16 - 17
Mild Thinness	17 - 18.5
Normal	18.5 - 25
Overweight	25 - 30
Obese Class I	30 - 35
Obese Class II	35 - 40
Obese Class III	> 40
```

```python
name = input("Enter your name: ")

weight = int(input("Enter your weight in kg: "))
height = float(input("Enter your height in m: "))

BMI = weight / (height ** 2)
print('Your BMI is: ', BMI)

if BMI > 0:
    if(BMI<16):
        print(name + " = Severe Thinness")
    elif(BMI <= 17):
        print(name + " = Moderate Thinness")
    elif(BMI <= 18.5):
        print(name + " = Mild Thinness")
    elif(BMI <= 25):
        print(name + " = Normal")
    elif(BMI <= 30):
        print(name + " = Overweight")
    elif(BMI < 35):
        print(name + " = Obese Class I")
    elif(BMI <= 40):
        print(name + " = Obese Class II")
    elif(BMI >= 40):
        print(name + " = Obese Class III")
else:
    print("Enter valid inputs")
```
